## 1. My rule and its reason codes

Rule:

Rank opportunities using a baseline score based on staleness and search volume. Higher staleness increases priority because stale pages may need refreshing, while higher search volume increases the potential value of the opportunity. The highest-scoring opportunities receive the corresponding action label.

Reason codes:

HIGH_STALENESS — the opportunity receives priority because its staleness signal is high.

HIGH_VOLUME — the opportunity receives priority because its search volume is high.

```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

print("Rows:", len(df))
print("Columns:", len(df.columns))
print(df.columns.tolist())
```

```python
# Create a simple baseline action score.
# Higher score = higher priority for action.

df["score"] = 0

# High impressions = enough search volume to justify an action
df.loc[df["impressions_90d"] >= 1000, "score"] += 2

# Position 5-20 = close enough to page 1 / strong ranking opportunity
df.loc[
    (df["avg_position"] >= 5) & (df["avg_position"] <= 20),
    "score"
] += 2

# Position 1-4 is already strong, so do not prioritize it
df.loc[df["avg_position"] == 0, "score"] = 0

# One reason code
df["reason_code"] = "NOT_PRIORITIZED"

df.loc[
    (df["impressions_90d"] >= 1000) &
    (df["avg_position"] >= 5) &
    (df["avg_position"] <= 20),
    "reason_code"
] = "HIGH_VOLUME_WEAK_POSITION"

# Action label
df["action"] = "MONITOR"

df.loc[
    df["reason_code"] == "HIGH_VOLUME_WEAK_POSITION",
    "action"
] = "QUICK_WIN"
```

```python
# Rank all content items by score.
# Tie-break with impressions so higher-volume items come first.

ranked = df.sort_values(
    ["score", "impressions_90d"],
    ascending=[False, False]
).reset_index(drop=True)

ranked["rank"] = ranked.index + 1

print("Total rows ranked:", len(ranked))
print(ranked[[
    "rank",
    "content_id",
    "client_id",
    "score",
    "reason_code",
    "action",
    "impressions_90d",
    "avg_position"
]].head(10))
```

```python
from pathlib import Path

OUTPUT_PATH = Path("../../work/outputs/baseline_action_score.csv")
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)

ranked.to_csv(OUTPUT_PATH, index=False)

print("CSV written to:", OUTPUT_PATH)
print("Rows written:", len(ranked))
```
