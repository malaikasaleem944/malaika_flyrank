## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

### Data Source and Scope

This analysis used the FlyRank ML Internship dataset. The work was based on the pseudonymized warehouse release `flyrank_pseudonymized_warehouse_release_v20260703`, with a snapshot export date of 2026-07-03.

The broader warehouse release contains four public-safe tables: `dim_clients`, `dim_content`, `fact_content_daily_performance`, and `fact_content_query_90d`. The production-scale release contains approximately 79 million rows in the daily content performance table.

For the modeling work in this capstone, I used the provided anonymized starter dataset containing 30,000 rows and 44 columns. The analysis used aggregated trailing-90-day performance measures, including `impressions_90d`, `clicks_90d`, `sessions_90d`, `content_age_days`, `avg_position`, `ctr`, and related trend fields.

Client names, URLs, page-level identifiers, and private search queries were excluded to keep the analysis public-safe. Fields directly derived from the prediction label were also excluded from the model features to reduce obvious leakage risk.


## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

### Methodology

#### Assumptions

This analysis assumes that the anonymized aggregated performance metrics provide useful signals for predicting future or target-window `clicks_90d`. The results are interpreted as decision-support evidence and not as a guarantee of business improvement or causal impact.

#### Features and Label

The prediction target was `clicks_90d`.

The model used the following features:

* `impressions_90d`
* `sessions_90d`
* `content_age_days`
* `avg_position`
* `ctr`
* `trend_pct`

Fields such as `trend_direction` and `is_declining_label` were excluded because they were identified as label-derived or potentially leakage-prone fields.

#### Model and Baseline

A Random Forest regression model was used as the primary model. Its performance was compared against a simple mean-prediction baseline. The baseline predicts the average target value from the training data for every observation in the test set.

Using a baseline ensures that the model is evaluated against a simple reference rather than being interpreted in isolation.

#### Validation Design

Two validation approaches were examined during the project: a random train/test split and a client-grouped split.

The final comparison focused on the client-grouped validation design, where clients in the test set were not present in the training set. This design provides a more relevant test of whether the model can generalize across previously unseen clients.

Model performance was measured using:

* Mean Absolute Error (MAE)
* Root Mean Squared Error (RMSE)
* R-squared (R²)

The Random Forest model and the mean-prediction baseline were evaluated on the same client-grouped test split.

#### Leakage Checks

A feature leakage audit was performed before final modeling. Known label-derived fields were identified and excluded from the feature set.

However, a remaining limitation is temporal leakage risk. Because the starter dataset contains rolling aggregate metrics, the exact timing relationship between every feature window and the target window cannot be fully reconstructed. Therefore, the analysis does not claim that the model is completely leakage-free. The findings should be interpreted as measured predictive evidence within the available dataset and validation design.


### Results

The final comparison used the same client-grouped test split for both the Random Forest model and the mean-prediction baseline. Clients in the test set were not included in the training set.

| Model                    |    MAE |   RMSE |     R² |
| ------------------------ | -----: | -----: | -----: |
| Mean-Prediction Baseline | 21.592 | 64.622 | -0.007 |
| Random Forest Regressor  |  1.134 | 23.509 |  0.867 |

The Random Forest model achieved lower MAE and RMSE than the baseline on the same client-grouped test split. It also achieved a substantially higher R² value.

Compared with the baseline, the measured results indicate that the selected aggregated performance features contained useful predictive signal within this evaluation setup. The result is directional and decision-support evidence; it does not demonstrate causal impact or guarantee equivalent performance in future production environments.

### Validation Interpretation

The grouped validation design is more challenging than a random split because the model is evaluated on clients that were not included during training. This makes the reported result more relevant to the specific question of cross-client generalization.

The results should still be interpreted with caution because the available starter dataset contains rolling aggregate features whose exact temporal alignment cannot be fully reconstructed. Therefore, the model should not be described as production-ready or fully leakage-free based on this analysis alone.

## 5. Limitations

*What this work cannot claim.*

### Limitations and Honest Framing

This work has several important limitations.

#### 1. Temporal leakage cannot be fully ruled out

The dataset contains rolling aggregate metrics, but the exact timing relationship between every feature window and the prediction target cannot be fully reconstructed from the available starter dataset. Therefore, this analysis cannot claim that all forms of temporal leakage have been eliminated.

#### 2. The evaluation is limited to the available dataset

The model was evaluated using a client-grouped split, which tests performance on clients not included in training. However, the results were measured within the available anonymized dataset and may not represent performance on completely new future data or production environments.

#### 3. Prediction does not demonstrate causation

The model identifies predictive relationships between the available features and `clicks_90d`. These relationships should not be interpreted as evidence that changing one feature will directly cause an increase or decrease in clicks.

#### 4. Aggregated features limit detailed interpretation

The analysis used public-safe aggregated metrics rather than client names, URLs, page-level identifiers, or private search queries. This protects privacy and confidentiality but limits page-specific investigation.

#### 5. Recommendations require human review

The ranked recommendations and action playbook are decision-support outputs. They are not automatic instructions and should be reviewed by a human before implementation.

Overall, this project presents measured and directional evidence from the available data. It does not claim guaranteed business improvement, causal effects, complete leakage elimination, or production-ready performance without further validation.

These recommendations are based on observed patterns in the available dataset. They should be treated as prioritization guidance rather than guaranteed optimization actions.
