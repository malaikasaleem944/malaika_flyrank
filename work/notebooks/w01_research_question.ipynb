## 3. Quick look at the data (2-3 real numbers)

### What I observed

The starter dataset contains **30,000 pages**. All **30,000 pages** in this starter slice have positive 90-day impressions, which means the pages have measurable search exposure.

There are **16,262 pages** with `trend_direction = "down"`, which is about **54.2%** of the starter dataset. This suggests that there is enough variation in the current performance signals to investigate how pages could be prioritized for review.

These numbers make the Refresh / Content Opportunity Scoring lane worth investigating. However, the `trend_direction` field is a current-window proxy, so I cannot treat these 16,262 pages as proof that they will decline in the future.


```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

total_pages = len(df)
declining_pages = (df["trend_direction"] == "down").sum()
pages_with_impressions = (df["impressions_90d"] > 0).sum()

print("Total pages:", total_pages)
print("Pages with impressions:", pages_with_impressions)
print("Pages with downward trend:", declining_pages)
```
